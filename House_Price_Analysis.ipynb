# Key Insights

1. The average house price is **₹540,088.64**.

2. Most houses in the dataset have **3 bedrooms (9,823 houses)**, followed by **4 bedrooms (6,882 houses)**.

3. House prices increase significantly with the **grade** of the house. Grade 13 houses have the highest average prices, while lower-grade houses have the lowest.

4. Waterfront properties are much more expensive. The average price of a waterfront house is **₹1,661,876**, compared to **₹531,564** for non-waterfront houses.

5. Houses with better views generally have higher average prices. Houses with **view = 4** have the highest average price (about **₹1.46 million**).

6. The dataset contains a few unusual entries, such as houses with **33 bedrooms**, indicating possible outliers or data entry errors.

7. The majority of houses do **not** have a waterfront, showing that waterfront properties are relatively rare.

8. The bar chart shows that **3-bedroom and 4-bedroom houses dominate the dataset**, making them the most common house types.

9. The visualizations indicate that premium features such as **higher grade, better view, and waterfront access** are associated with higher house prices.

10. Overall, house prices are influenced by multiple factors, including **bedrooms, grade, view, and waterfront**, rather than a single feature.
